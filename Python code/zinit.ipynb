```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

valuesList = [1]

for value in valuesList:

    # -------------------------
    # SETTINGS
    # -------------------------
    outdir = "ED_Data/zAlignement"
    i0 = 1
    t_max_plot = 10
    alpha = 0.1
    N = 14
    Jz = 1
    Jx = 5

    vmin_corr = 1e-4
    prefix = "b1z"

    # -------------------------
    # LOAD DATA (corr only)
    # -------------------------
    df_corr = pd.read_csv(f"{outdir}/{prefix}corr.csv")

    # -------------------------
    # TIME AXIS
    # -------------------------
    t = df_corr["t"].values
    mask = t <= t_max_plot
    t_plot = t[mask]
    t_min, t_max = t_plot.min(), t_plot.max()

    # -------------------------
    # CONNECTED CORRELATIONS (Czz)
    # -------------------------
    C_cols = [c for c in df_corr.columns if c.startswith(f"Cxx_{i0}_")]
    j_vals = np.array([int(c.split("_")[-1]) for c in C_cols])
    order = np.argsort(j_vals)

    CorrMat = np.abs(df_corr[C_cols].to_numpy()[:, order])
    CorrMat = np.clip(CorrMat, vmin_corr, 1.0)
    CorrMat = CorrMat[mask, :]

    # -------------------------
    # SINGLE PLOT (corr only)
    # -------------------------
    plt.figure(figsize=(5, 3))

    extent = [0.5, CorrMat.shape[1] + 0.5, t_min, t_max]

    im = plt.imshow(
        CorrMat,
        origin="lower",
        aspect="auto",
        cmap="turbo",
        interpolation="nearest",
        norm=mcolors.LogNorm(vmin=vmin_corr, vmax=1),
        extent=extent,
    )

    plt.colorbar(im, label=r"$|C_{2,j}^{xx}(t)|$")
    plt.xlabel(r"Site $j$")
    plt.ylabel(r"Time $t$")

    if alpha >= 100:
        plt.title(rf"$N={N}$, $J_z={Jz}$, $J_x={Jx}$, $\alpha\to\infty$")
    else:
        plt.title(
            rf"$|1011\cdots\rangle$ "
            rf"($N={N}$, $J_z={Jz}$, $J_x={Jx}$, $\alpha={alpha}$)"
        )



    plt.tight_layout()
    plt.savefig("plt/zinit.svg")
    plt.show()
```
