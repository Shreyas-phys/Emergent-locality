```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import numpy as np

JxValues=[0.1, 1, 5, 10, 20]

print(os.getcwd())

relpath=r"ED_Data/Trial/"
prefix="b2Simple"

df=pd.read_csv(f"{relpath}{prefix}corrzz.csv")
heatmap = df.to_numpy()

plt.figure(figsize=(5, 2.5))

plt.imshow(
    np.abs(heatmap[:, 1:]),
    origin="lower",
    aspect="auto",
    cmap="turbo",
    interpolation="nearest",
    norm=mcolors.LogNorm(vmin=1e-10, vmax=1),
)

plt.xlabel("Site index")
plt.ylabel("Time index")

nx = heatmap[:, 1:].shape[1]
step = max(1, nx // 6)

ny = heatmap[:, 1:].shape[0]

step_y = max(1, ny // 10)
plt.xticks(
    ticks=np.arange(0, nx, step),
    labels=np.arange(1, nx+1, step)
)

plt.yticks(
    ticks = np.linspace(0, ny - 1, 11).astype(int),
    labels = np.linspace(0, 100, 11).astype(int)
)

plt.colorbar(label="Corr")
plt.tight_layout()
plt.show()

```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import numpy as np

# Jx = 10  # example value for this plot
JxValues=[0.1, 5,10,20, 100]

for i, Jx in enumerate(JxValues):
    relpath = r"ED_Data/VaryJx/"
    prefix = f"LRb2Jx{Jx}"

    df = pd.read_csv(f"{relpath}{prefix}corr.csv")
    heatmap = df.to_numpy()

    plt.figure(figsize=(5, 2.5))

    plt.imshow(
        np.abs(heatmap[:, 1:]),
        origin="lower",
        aspect="auto",
        cmap="turbo",
        interpolation="nearest",
        norm=mcolors.LogNorm(vmin=1e-5, vmax=1),
    )

    # ---- title ----
    plt.title(rf"$(2,11;3,11)\;\; J_x = {Jx}, J_z=1, \alpha=0.2$", fontsize=11)

    # ---- axis labels ----
    plt.xlabel("site")
    plt.ylabel("time")

    # ---- ticks ----
    nx = heatmap[:, 1:].shape[1]
    step_x = max(1, nx // 6)

    ny = heatmap[:, 1:].shape[0]

    plt.xticks(
        ticks=np.arange(0, nx, step_x),
        labels=np.arange(1, nx+1, step_x)
    )

    plt.yticks(
        ticks=np.linspace(0, ny - 1, 6).astype(int),
        labels=np.linspace(0, 10, 6).astype(int)
    )

    # ---- panel letter (inside axes) ----
    panel_label = f"({chr(ord('a') + i)})"
    plt.text(
        0.02, 0.95,
        panel_label,
        transform=plt.gca().transAxes,
        fontsize=11,
        fontweight="bold",
        va="top",
        bbox=dict(
            facecolor="white",
            edgecolor="none",
            alpha=0.8
        )
    )


    plt.colorbar(label="Corr")
    plt.tight_layout()
    plt.show()
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

relpath = r"ED_Data/Projected/"

# ----------------------------
# Suggested initial conditions
# ----------------------------
configs = [
    # (prefix_used_in_run, theta, phi, display_label)
    ("initvary1",   np.pi/2, 0.0,     r"$\theta=\pi/2,\ \phi=0$"),
    ("initvary2",  np.pi/2, np.pi,   r"$\theta=\pi/2,\ \phi=\pi$"),
    ("initvary3", np.pi/2, np.pi/2, r"$\theta=\pi/2,\ \phi=\pi/2$"),
]

def plot_conc_heatmap(ax, csv_path, title):
    df = pd.read_csv(csv_path)
    heatmap = df.to_numpy()

    im = ax.imshow(
        np.abs(heatmap[:, 1:]),
        origin="lower",
        aspect="auto",
        cmap="turbo",
        interpolation="nearest",
        norm=mcolors.LogNorm(vmin=1e-10, vmax=1),
    )

    ax.set_title(title, fontsize=10)
    ax.set_xlabel("Site index")
    ax.set_ylabel("Time index")

    nx = heatmap[:, 1:].shape[1]
    step = max(1, nx // 6)
    ax.set_xticks(np.arange(0, nx, step))
    ax.set_xticklabels(np.arange(1, nx + 1, step))

    ny = heatmap[:, 1:].shape[0]
    ax.set_yticks(np.linspace(0, ny - 1, 11).astype(int))
    ax.set_yticklabels(np.linspace(0, 100, 11).astype(int))

    return im

print("Working dir:", os.getcwd())

fig, axes = plt.subplots(1, 3, figsize=(12, 3.2), constrained_layout=True)

im_last = None
for ax, (prefix, theta, phi, label) in zip(axes, configs):
    csv_path = os.path.join(relpath, f"{prefix}conc.csv")
    title = f"{prefix}\n{label}"
    im_last = plot_conc_heatmap(ax, csv_path, title)

# One shared colorbar for all subplots
cbar = fig.colorbar(im_last, ax=axes, shrink=0.95, pad=0.02)
cbar.set_label("Concurrence")

plt.show()
```
